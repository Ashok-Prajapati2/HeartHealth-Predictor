```python
import pandas as pd

# Creating the dataset as a dictionary
data = {
    "EmployeeID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Age": [25, 30, 35, 28, 45],
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Salary": [50000, 75000, 80000, 62000, 52000],
    "JoiningDate": ["2019-06-20", "2018-03-14", "2017-11-22", "2020-01-10", "2010-09-15"],
    "City": ["New York", "San Francisco", "Los Angeles", "Chicago", "New York"],
    "Experience": [4, 6, 7, 3, 14],
}

# Converting the dictionary into a DataFrame
df = pd.DataFrame(data)

# Saving the DataFrame to a CSV file
file_path = "employees.csv"
df.to_csv(file_path, index=False)

file_path
```
